```python
# Scenario
# A student is building a simple AI assistant in Google Colab for a class project. The assistant is
# hosted on Azure Foundry, and the student wants to ask it questions like:

# “What can you help with?”

# “Explain this topic simply.”

# “Give me a short summary.”

# Instead of talking directly to the model through a chat app, the student uses Python code to connect to the Azure service.

from google.colab import userdata

api_key = userdata.get("foundry_api")
print("Loaded key:", api_key[:6] + "..." if api_key else "Missing")

from google.colab import userdata
from openai import OpenAI

my_endpoint = "https://archanafoundry.services.ai.azure.com/api/projects/proj-default"
my_agent = "email-assistant"
my_version = "1"

api_key = userdata.get("foundry_api")

openai_client = OpenAI(
    base_url=f"{my_endpoint}/openai/v1",
    api_key=api_key
)

response = openai_client.responses.create(
    input=[{"role": "user", "content": "Tell me what you can help with."}],
    extra_body={
        "agent_reference": {
            "name": my_agent,
            "version": my_version,
            "type": "agent_reference"
        }
    },
)

print(response.output_text)
```

```python
#AI Interview Simulator with Feedback

#Concept: Simulates real interviews

#💡 Output:

#Questions
#Feedback on answers
#Improvement tips

#Twist:
#Add HR + Technical + Stress interview modes


from google.colab import userdata
from openai import OpenAI

api_key = userdata.get("foundry_api")

my_endpoint = "https://archanafoundry.services.ai.azure.com/api/projects/proj-default"
my_agent = "ai-interview-simulato"
my_version = "1"

client = OpenAI(
    base_url=f"{my_endpoint}/openai/v1",
    api_key=api_key
)

print("🤖 Chatbot Ready (type 'exit' to stop)\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Chatbot: Thank you! 😊")
        break

    try:
        response = client.responses.create(
            input=[
                {
                    "role": "user",
                    "content": user_input
                }
            ],
            extra_body={
                "agent_reference": {
                    "name": my_agent,
                    "version": my_version,
                    "type": "agent_reference"
                }
            },
        )

        print("Chatbot:", response.output_text)
        print("-" * 50)

    except Exception as e:
        print("Error:", str(e))
```
